# Roman Pointing Demonstration Notebook

This notebook follows the pointing model described in Savransky's Roman Space Telescope pointing paper. We simulate how to compute Roman's zero-point orientation, then apply yaw and pitch rotations to align the telescope boresight (x̂) with a target in space. We also compute the Sun angle to verify thermal constraints.

```python
import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u
from astropy.time import Time
from astropy.coordinates import SkyCoord, get_body_barycentric, BarycentricMeanEcliptic
from scipy.optimize import fsolve

# Helper rotation matrix functions
def rot_x(theta):
    return np.array([[1, 0, 0],
                     [0, np.cos(theta), np.sin(theta)],
                     [0, -np.sin(theta), np.cos(theta)]])

def rot_y(theta):
    return np.array([[np.cos(theta), 0, -np.sin(theta)],
                     [0, 1, 0],
                     [np.sin(theta), 0, np.cos(theta)]])

def rot_z(theta):
    return np.array([[np.cos(theta), np.sin(theta), 0],
                     [-np.sin(theta), np.cos(theta), 0],
                     [0, 0, 1]])
```

## Compute L2 Location
We approximate the position of the Earth-Sun L2 point using a collinear Lagrange point solver.

```python
# Compute mass ratio for Earth-Sun system
M_earth = 1 * u.Mearth
M_sun = 1 * u.Msun
mu = (M_earth / (M_earth + M_sun)).decompose().value

# Function to solve for L2 location
def lagrange_eq(x):
    return x - (1 - mu)*(x + mu)/abs(x + mu)**3 - mu*(x - 1 + mu)/abs(x - 1 + mu)**3

L2_loc = fsolve(lagrange_eq, 1)[0]
L2_loc
```
